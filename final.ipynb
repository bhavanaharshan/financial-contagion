UNDERSTANDING THE TABLE

- L_REP_CTY: Reporting Country or Source node - this is the lender
- L_CP_COUNTRY: Counterparty Country or Target node - this is the receiver
- quarter: time stamp in which these transactions are occurring
- exposure: edge weight which is the total value of financial claim (in millions of US dollars)

Here, if country 1C goes bankrupt, country 5A is at the risk of losing 3,562 million dollars

NEXT STEPS:
- Node mapping: String to Integer (create a dictionary) and this must be consistent
- Normalisation of exposure values : Apply log normalisation i.e., Weight = log (Exposure + 1)
- Graph Construction: Table to Graph Objects 
- For each quarter, we need a Pytorch Geometric Data object containing
   - edge_index: who connects to whom
   - edge_attr: the normalized exposure amount
   - x: node features which is a vector of 7: [Page rank, Betweenness, Closeness, In-Degree, Out-Degree, Net position, Clustering]

```python
import pandas as pd
import numpy as np
import networkx as nx
import torch
from torch_geometric.data import Data 
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("cleaned_lbs_exposures.csv")
df["quarter"] = df["quarter"].astype("str")
print("Data loaded, total rows: ", len(df))

df.head()
```

```python
# Node mapping
# Collect all unique countries appearing in either Source or Target and sort them
# Assign ID and create dictionary

df.loc[df['exposure'] < 0, 'exposure'] = 0.0
sources = df['L_REP_CTY'].unique()
targets = df['L_CP_COUNTRY'].unique()

all_countries = pd.unique(np.concatenate((sources, targets), axis=0))
all_countries.sort()

node_map = {code: i for i, code in enumerate(all_countries)}
num_nodes = len(node_map)

print(f"Total Unique Countries (Nodes): {num_nodes}")
print("First 5 mappings:", {k: node_map[k] for k in list(node_map)[:5]})
```

```python
# Build Graph for a specific point in time
# Metric Calculation (exposure used for centrality metrics)

def calculate_sna_features(df_quarter, all_nodes_map):
    G = nx.DiGraph()
    G.add_nodes_from(all_nodes_map.keys())
    
    edges = list(zip(df_quarter['L_REP_CTY'], df_quarter['L_CP_COUNTRY'], df_quarter['exposure']))
    G.add_weighted_edges_from(edges)
    
    # PAGE RANK
    try:
        pr = nx.pagerank(G, weight='weight', alpha=0.85)
    except:
        # Fallback for convergence errors
        pr = {n: 0 for n in G.nodes()}
        
    # BETWEENNESS 
    bet = nx.betweenness_centrality(G, weight=None)
    
    # CLOSENESS
    clo = nx.closeness_centrality(G)
    
    # IN-DEGREE
    in_deg = dict(G.in_degree(weight='weight'))
    
    # OUT-DEGREE
    out_deg = dict(G.out_degree(weight='weight'))
    
    # NET POSITION
    net_pos = {n: out_deg[n] - in_deg[n] for n in G.nodes()}
    
    # CLUSTERING COEFFICIENT
    clust = nx.clustering(G)
    
    sorted_node_codes = sorted(all_nodes_map.keys(), key=lambda x: all_nodes_map[x])
    
    feature_matrix = []
    for code in sorted_node_codes:
        vec = [
            pr.get(code, 0),
            bet.get(code, 0),
            clo.get(code, 0),
            in_deg.get(code, 0),
            out_deg.get(code, 0),
            net_pos.get(code, 0),
            clust.get(code, 0)
        ]
        feature_matrix.append(vec)
        
    return np.array(feature_matrix)
```

NEXT: Graph Construction Loop

- go through each unique quarter chronologically
- call SNA function to get feature matrix
- feature normalization (pagerank can be between 0 and 1, while net position can be in billions) : Apply StandardScaler

- edge normalisation which is earlier mentioned log normalization
- pack edge_index, edge_attr and x into Data object

```python
snapshots = []
quarters = sorted(df["quarter"].unique())

print(f"Processing {len(quarters)} quarters..")

for q in quarters:
    
    df_q = df[df["quarter"] == q]
    
    # Generate Node features (X)
    sna_features = calculate_sna_features(df_q, node_map)
    scaler = StandardScaler()
    sna_features_norm = scaler.fit_transform(sna_features)
    
    x = torch.tensor(sna_features_norm, dtype= torch.float)
    
    # Generate Edges (Edge index)
    src = df_q['L_REP_CTY'].map(node_map).values
    dst = df_q['L_CP_COUNTRY'].map(node_map).values
    edge_index = torch.tensor(np.array([src, dst]), dtype=torch.long)
    
    # Generate Edge weights (edge_attr) with normalization
    weights_log = np.log1p(df_q['exposure'].values)
    edge_attr = torch.tensor(weights_log, dtype=torch.float).view(-1, 1)
    
    
    data = Data(x=x, edge_index=edge_index, edge_attr=edge_attr)
    snapshots.append(data)

print("Construction Complete.")
torch.save(snapshots, "sna_graph_snapshots.pt")
torch.save(node_map, "node_mapping.pt")   
    
    
```

STATIC MODEL: 
- Input: takes one graph snapshot at a time. It sees the countries, their connections and their SNA metrics
- Uses GCNConv to aggregate information from neighbours. 
- Model outputs a learned (embedding) vector for each country
- Calculates dot product between 2 countries

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv

class StaticGCN(torch.nn.Module):
    def __init__(self, num_features, hidden_dim, output_dim):
        super(StaticGCN, self).__init__()
        # Layer 1: Input (7 SNA metrics) -> Hidden Layer
        self.conv1 = GCNConv(num_features, hidden_dim)
        
        # Layer 2: Hidden -> Output Embedding (Risk Vector)
        self.conv2 = GCNConv(hidden_dim, output_dim)

    def forward(self, data):
        x, edge_index, edge_weight = data.x, data.edge_index, data.edge_attr
        
        # --- First Graph Convolution ---
        # We use the edge weights (exposure amounts) to guide the aggregation
        x = self.conv1(x, edge_index, edge_weight)
        x = F.relu(x)
        x = F.dropout(x, p=0.2, training=self.training) # Prevent overfitting
        
        # --- Second Graph Convolution ---
        x = self.conv2(x, edge_index, edge_weight)
        
        # We don't apply Softmax here because we need the raw vector 
        # for the Dot Product step later.
        return x

print("Static GCN Model defined.")
```

```python
import torch.optim as optim
from sklearn.metrics import roc_auc_score

# 1. Setup
snapshots = torch.load("sna_graph_snapshots.pt", weights_only=False)
node_map = torch.load("node_mapping.pt")

# Hyperparameters
HIDDEN_DIM = 32
EMBEDDING_DIM = 16 # Size of the "Risk Vector"
LEARNING_RATE = 0.01
EPOCHS = 30 # Short training for demonstration (increase for real results)

# Initialize Model
model = StaticGCN(num_features=7, hidden_dim=HIDDEN_DIM, output_dim=EMBEDDING_DIM)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
criterion = torch.nn.BCEWithLogitsLoss() # Binary Cross Entropy

# Split Index (Same as before)
split_idx = int(len(snapshots) * 0.8)

print(f"Starting Training on {split_idx} snapshots...")

# --- Training Loop ---
for epoch in range(EPOCHS):
    total_loss = 0
    model.train()
    
    # Iterate through TIME (Trajectory)
    # We stop 1 step early because we need t+1 for labels
    for t in range(split_idx - 1):
        curr_graph = snapshots[t]
        next_graph = snapshots[t+1]
        
        optimizer.zero_grad()
        
        # 1. Forward Pass
        # Get Node Embeddings (Risk Vectors) for time t
        z = model(curr_graph)
        
        # 2. Calculate "Risk Scores" for existing edges (Dot Product)
        # We only care about edges that exist now (to see if they fail)
        src, dst = curr_graph.edge_index
        
        # Dot Product: (Embed_src * Embed_dst).sum(dim=1)
        # This measures alignment/risk flow
        score = (z[src] * z[dst]).sum(dim=1)
        
        # 3. Generate Ground Truth Labels (On the fly)
        # Did exposure drop > 10% in next_graph?
        
        # Map current edges to future weights
        # (Optimized lookup using tensors roughly)
        curr_w = curr_graph.edge_attr.flatten()
        
        # Create a sparse matrix logic or dictionary for fast lookup
        # (Using dictionary here for clarity/safety)
        next_edges_set = set(zip(next_graph.edge_index[0].tolist(), next_graph.edge_index[1].tolist()))
        next_w_dict = {
            (u.item(), v.item()): w.item() 
            for u, v, w in zip(next_graph.edge_index[0], next_graph.edge_index[1], next_graph.edge_attr)
        }
        
        labels = []
        for i in range(len(src)):
            u, v = src[i].item(), dst[i].item()
            c_w = curr_w[i].item()
            
            # Find future weight
            n_w = next_w_dict.get((u, v), 0.0) # 0 if edge disappears
            
            # Calc Drop
            if c_w > 0:
                pct_change = (n_w - c_w) / c_w
            else:
                pct_change = 0
            
            # Risk = 1 if drop < -0.10
            labels.append(1.0 if pct_change < -0.10 else 0.0)
            
        labels_tensor = torch.tensor(labels)
        
        # 4. Backprop
        loss = criterion(score, labels_tensor)
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        
    if (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1}/{EPOCHS} | Avg Loss: {total_loss / split_idx:.4f}")

print("Static GCN Training Complete.")
```

```python
import pandas as pd

# Reverse map for naming
id_to_country = {v: k for k, v in node_map.items()}

results_rows = []
model.eval()

print("Generating Rankings for Test Set (Future)...")

with torch.no_grad():
    for t in range(split_idx, len(snapshots) - 1):
        curr_graph = snapshots[t]
        
        # 1. Get Embeddings
        z = model(curr_graph)
        
        # 2. Score Edges
        src, dst = curr_graph.edge_index
        
        # Dot Product Score (The higher, the more "aligned" for risk)
        raw_scores = (z[src] * z[dst]).sum(dim=1)
        
        # Convert to probability (Sigmoid) for easier reading
        probs = torch.sigmoid(raw_scores).numpy()
        
        # 3. Store Results
        for i in range(len(src)):
            u_id = src[i].item()
            v_id = dst[i].item()
            prob = probs[i]
            
            results_rows.append({
                'Source': id_to_country[u_id],
                'Target': id_to_country[v_id],
                'time_idx': t,
                'Static_GNN_Prob': prob
            })

# Convert to DataFrame
df_static = pd.DataFrame(results_rows)

# Create Rank: 1 = Highest Risk
df_static['Static_Rank'] = df_static['Static_GNN_Prob'].rank(ascending=False).astype(int)

# Filter columns
final_static_table = df_static[['Source', 'Target', 'time_idx', 'Static_Rank', 'Static_GNN_Prob']].sort_values('Static_Rank')

print("\nTop 10 Riskiest Pathways predicted by Static GCN:")
print(final_static_table.head(10))

# Save
df_static.to_csv("results_static_gnn.csv", index=False)
print("Saved 'results_static_gnn.csv'")
```

DYNAMIC GNN:

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv

class DynamicGNN(torch.nn.Module):
    def __init__(self, num_nodes, num_features, hidden_dim, output_dim):
        super(DynamicGNN, self).__init__()
        
        # 1. Spatial Engine (GCN)
        # Compresses the 7 SNA metrics + Neighbor info into a "Context Vector"
        self.gcn = GCNConv(num_features, hidden_dim)
        
        # 2. Temporal Engine (GRU Cell)
        # Input: The context vector from GCN
        # Hidden: The memory from the previous step
        self.gru = torch.nn.GRUCell(hidden_dim, hidden_dim)
        
        # 3. Output Layer
        # Projects the GRU memory to the final Risk Vector
        self.linear = torch.nn.Linear(hidden_dim, output_dim)

    def forward(self, data, h_prev):
        """
        data: The graph snapshot at time t
        h_prev: The hidden state (memory) from time t-1
        """
        x, edge_index, edge_weight = data.x, data.edge_index, data.edge_attr
        
        # --- A. Spatial Step (GCN) ---
        # "What is happening right now in the network?"
        x_spatial = self.gcn(x, edge_index, edge_weight)
        x_spatial = F.relu(x_spatial)
        
        # --- B. Temporal Step (GRU) ---
        # "Combine right now with what I remember from the past"
        # h_new shape: [Num_Nodes, Hidden_Dim]
        h_new = self.gru(x_spatial, h_prev)
        
        # --- C. Output Step ---
        # Create the final Risk Vector
        out = self.linear(h_new)
        
        return out, h_new

print("Dynamic GNN (GCN + GRU) Architecture Defined.")
```

```python
import torch.optim as optim

# Setup
snapshots = torch.load("sna_graph_snapshots.pt", weights_only=False)
num_nodes = snapshots[0].num_nodes
split_idx = int(len(snapshots) * 0.8)

# Hyperparameters
HIDDEN_DIM = 32
EMBEDDING_DIM = 16 
LEARNING_RATE = 0.01
EPOCHS = 30

# Initialize Model
model_dyn = DynamicGNN(num_nodes=num_nodes, num_features=7, 
                       hidden_dim=HIDDEN_DIM, output_dim=EMBEDDING_DIM)

optimizer = optim.Adam(model_dyn.parameters(), lr=LEARNING_RATE)
criterion = torch.nn.BCEWithLogitsLoss()

print(f"Starting Dynamic Training on {split_idx} snapshots...")

for epoch in range(EPOCHS):
    total_loss = 0
    model_dyn.train()
    
    # Initialize Hidden State (Memory) with Zeros for all nodes
    h = torch.zeros(num_nodes, HIDDEN_DIM)
    
    # Iterate through TIME
    for t in range(split_idx - 1):
        curr_graph = snapshots[t]
        next_graph = snapshots[t+1]
        
        optimizer.zero_grad()
        
        # 1. Forward Pass
        # We pass the graph AND the previous memory 'h'
        z, h = model_dyn(curr_graph, h)
        
        # IMPORTANT: Detach h from the computation graph to prevent memory explosion
        # This means we only backpropagate errors for 1 step (Truncated BPTT)
        h = h.detach()
        
        # 2. Risk Scoring (Dot Product)
        src, dst = curr_graph.edge_index
        score = (z[src] * z[dst]).sum(dim=1)
        
        # 3. Ground Truth Generation (Same logic as Static)
        # (Did exposure drop > 10%?)
        curr_w = curr_graph.edge_attr.flatten()
        
        # Lookup table for next weights
        next_w_dict = {
            (u.item(), v.item()): w.item() 
            for u, v, w in zip(next_graph.edge_index[0], next_graph.edge_index[1], next_graph.edge_attr)
        }
        
        labels = []
        for i in range(len(src)):
            u, v = src[i].item(), dst[i].item()
            c_w = curr_w[i].item()
            n_w = next_w_dict.get((u, v), 0.0)
            
            if c_w > 0: pct_change = (n_w - c_w) / c_w
            else: pct_change = 0
            
            labels.append(1.0 if pct_change < -0.10 else 0.0)
            
        labels_tensor = torch.tensor(labels)
        
        # 4. Backprop
        loss = criterion(score, labels_tensor)
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()

    if (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1}/{EPOCHS} | Avg Loss: {total_loss / split_idx:.4f}")

print("Dynamic GNN Training Complete.")
```

```python
import pandas as pd

results_rows = []
model_dyn.eval()

# Reverse map
id_to_country = {v: k for k, v in torch.load("node_mapping.pt").items()}

# We continue from the last 'h' state of training (The model remembers the past!)
# Ideally, we run through the whole sequence again to get the state up to split_idx
# For simplicity here, we assume h is warmed up or reset. 
# Let's reset to be safe for a clean evaluation run.
h = torch.zeros(num_nodes, HIDDEN_DIM)

print("Generating Rankings for Test Set (Dynamic)...")

with torch.no_grad():
    for t in range(len(snapshots) - 1): # Run through whole history
        curr_graph = snapshots[t]
        
        # Forward pass
        z, h = model_dyn(curr_graph, h)
        
        # Only save results if we are in the TEST period
        if t >= split_idx:
            src, dst = curr_graph.edge_index
            raw_scores = (z[src] * z[dst]).sum(dim=1)
            probs = torch.sigmoid(raw_scores).numpy()
            
            for i in range(len(src)):
                results_rows.append({
                    'Source': id_to_country[src[i].item()],
                    'Target': id_to_country[dst[i].item()],
                    'time_idx': t,
                    'Dynamic_GNN_Prob': probs[i]
                })

# Create DataFrame
df_dyn = pd.DataFrame(results_rows)
df_dyn['Dynamic_Rank'] = df_dyn['Dynamic_GNN_Prob'].rank(ascending=False).astype(int)

# Save
df_dyn.to_csv("results_dynamic_gnn.csv", index=False)
print("Saved 'results_dynamic_gnn.csv'")

print("\nTop 10 Riskiest Pathways predicted by Dynamic GNN:")
print(df_dyn[['Source', 'Target', 'time_idx', 'Dynamic_Rank']].sort_values('Dynamic_Rank').head(10))
```

```python
import pandas as pd

# 1. Load the Model Results
df_dyn = pd.read_csv("results_dynamic_gnn.csv")
df_static = pd.read_csv("results_static_gnn.csv")

# 2. Get the "Ground Truth" Labels
# We need to know which edges actually crashed (Label = 1).
# We pull this from the previous file, but we ignore the SNA predictions.
df_labels = pd.read_csv("ground_truth.csv")[['Source', 'Target', 'time_idx', 'label']]

# 3. Merge the Three DataFrames
# We align them perfectly on Source, Target, and Time
df_final = df_dyn.merge(
    df_static[['Source', 'Target', 'time_idx', 'Static_Rank']],
    on=['Source', 'Target', 'time_idx'],
    how='inner'
).merge(
    df_labels,
    on=['Source', 'Target', 'time_idx'],
    how='inner'
)

# 4. Filter: Look ONLY at Actual Risk Events
# We don't care about safe edges. We want to know: "When a crash happened, who saw it coming?"
df_risks = df_final[df_final['label'] == 1].copy()

# 5. Define Detection Logic
# If Rank is in the Top 500, we say it was "Detected". Otherwise, it was "Missed".
THRESHOLD = 500

df_risks['Dynamic Status'] = df_risks['Dynamic_Rank'].apply(lambda x: "DETECTED" if x <= THRESHOLD else "MISSED")
df_risks['Static Status'] = df_risks['Static_Rank'].apply(lambda x: "DETECTED" if x <= THRESHOLD else "MISSED")

# 6. Create "Performance Gap" Metric
# How many times better was the Dynamic Model?
# Ratio = Static Rank / Dynamic Rank. 
# (e.g., Static Rank 100,000 / Dynamic Rank 1 = 100,000x improvement)
df_risks['Improvement Factor'] = (df_risks['Static_Rank'] / df_risks['Dynamic_Rank']).round(1)

# 7. Create Readable "Pathway" Column
df_risks['Risk Pathway'] = df_risks['Source'] + " -> " + df_risks['Target']

# 8. Final Table Formatting
evidence_table = df_risks[[
    'Risk Pathway', 
    'time_idx',
    'Dynamic_Rank', 'Dynamic Status',
    'Static_Rank', 'Static Status',
    'Improvement Factor'
]].sort_values('Dynamic_Rank')

# Print
print("=============================================================================")
print("FINAL EVIDENCE: DYNAMIC GNN (Spatiotemporal) vs. STATIC GNN (Spatial)")
print(f"Metrics: Ranks represent priority. Lower is better. Threshold = {THRESHOLD}")
print("=============================================================================")
print(evidence_table.head(15).to_string(index=False))

# Save for your report
evidence_table.to_csv("final_evidence_dynamic_vs_static.csv", index=False)
```

EVALUATION METRICS
- how can I know dynamic and static predicted correctly?

```python
import pandas as pd
import torch
import numpy as np

# 1. Load Data
df_dyn = pd.read_csv("results_dynamic_gnn.csv")
df_static = pd.read_csv("results_static_gnn.csv")
snapshots = torch.load("sna_graph_snapshots.pt", weights_only=False)
node_map = torch.load("node_mapping.pt")

# Helper to reverse map
id_to_country = {v: k for k, v in node_map.items()}

# 2. Pick the Top Risk found by Dynamic GNN
# We look for the row with Dynamic_Rank == 1 in the last available time step
top_risk = df_dyn.sort_values("Dynamic_Rank").iloc[0]

src_country = top_risk['Source']
dst_country = top_risk['Target']
time_idx = top_risk['time_idx']

print(f"=== AUDITING TOP PREDICTION: {src_country} -> {dst_country} at T={time_idx} ===")

# 3. Get the ACTUAL Dollar Amounts from the Snapshots
# We need to find the node IDs first
src_id = node_map[src_country]
dst_id = node_map[dst_country]

# Function to get weight
def get_exposure(graph, u, v):
    # Find the index of the edge u->v
    edge_indices = graph.edge_index
    for i in range(edge_indices.shape[1]):
        if edge_indices[0, i] == u and edge_indices[1, i] == v:
            # We stored log(exposure+1), so we reverse it: exp(val) - 1
            log_val = graph.edge_attr[i].item()
            return np.expm1(log_val)
    return 0.0 # Edge doesn't exist

# Get Current and Future Exposure
current_exposure = get_exposure(snapshots[time_idx], src_id, dst_id)
future_exposure = get_exposure(snapshots[time_idx + 1], src_id, dst_id)

# 4. Calculate the REAL drop
if current_exposure > 0:
    pct_change = (future_exposure - current_exposure) / current_exposure
else:
    pct_change = 0.0

print(f"\n--- THE REALITY (Ground Truth) ---")
print(f"Exposure at Time {time_idx}:     ${current_exposure:,.2f} Million")
print(f"Exposure at Time {time_idx+1}:   ${future_exposure:,.2f} Million")
print(f"ACTUAL CHANGE:               {pct_change*100:.2f}%")

if pct_change < -0.10:
    print("VERDICT: CRASH CONFIRMED. (Label = 1)")
else:
    print("VERDICT: NO CRASH. (Label = 0)")

# 5. Compare Predictions
print(f"\n--- THE PREDICTIONS ---")

# Get Static Rank for this specific edge/time
static_row = df_static[
    (df_static['Source'] == src_country) & 
    (df_static['Target'] == dst_country) & 
    (df_static['time_idx'] == time_idx)
]
static_rank = static_row['Static_Rank'].values[0] if not static_row.empty else "N/A"

print(f"Dynamic Rank: {top_risk['Dynamic_Rank']} (Predicted High Risk)")
print(f"Static Rank:  {static_rank} (Predicted Safe/Ignore)")

print("\n--- CONCLUSION ---")
if top_risk['Dynamic_Rank'] < 500 and pct_change < -0.10:
    print("The Dynamic Model was CORRECT. It foresaw the crash.")
else:
    print("The Dynamic Model was INCORRECT.")

if static_rank > 500 and pct_change < -0.10:
    print("The Static Model was INCORRECT. It missed the crash completely.")
```

- Detection success rate (precision@k): how many are actual crashes?

```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
import torch
import numpy as np

sns.set_theme(style="darkgrid", context='notebook')
plt.rcParams.update({'font.size':12})

df_dyn = pd.read_csv("results_dynamic_gnn.csv")
df_static = pd.read_csv("results_static_gnn.csv")
df_labels = pd.read_csv("ground_truth.csv")[['Source', 'Target', 'time_idx', 'label']]

df_merged = df_dyn.merge(
    df_static[['Source', 'Target', 'time_idx', 'Static_Rank']],
    on=['Source', 'Target', 'time_idx']
).merge(df_labels, on=['Source', 'Target', 'time_idx'])

df_crashes = df_merged[df_merged['label'] == 1].copy()

top_5_risks = df_crashes.sort_values('Dynamic_Rank').head(5).copy()
top_5_risks['Pathway'] = top_5_risks['Source'] + " -> " + top_5_risks['Target']

plt.figure(figsize=(10, 6))

x = np.arange(len(top_5_risks))
width = 0.35

plt.bar(x - width/2, top_5_risks['Dynamic_Rank'], width, label='Dynamic GNN Rank', color='#2ecc71')
plt.bar(x + width/2, top_5_risks['Static_Rank'], width, label='Static GNN Rank (Baseline)', color='#e74c3c')

plt.yscale('log')
plt.ylabel('Rank (Lower is Better) [Log Scale]')
plt.title('Ranking Comparison on Top 5 Actual Contagion Events')
plt.xticks(x, top_5_risks['Pathway'])
plt.legend()
plt.tight_layout()
plt.savefig('graph_1_ranking_gap.png')
print("Graph 1 Saved: 'graph_1_ranking_gap.png'")

snapshots = torch.load("sna_graph_snapshots.pt", weights_only=False)
node_map = torch.load("node_mapping.pt")
src_id = node_map['IM']
dst_id = node_map['MD']

# Extract history for T=175 to T=185 (around the crash at 182)
history_times = range(175, 185)
history_values = []

for t in history_times:
    # Find weight
    val = 0.0
    if t < len(snapshots):
        edge_indices = snapshots[t].edge_index
        for i in range(edge_indices.shape[1]):
            if edge_indices[0, i] == src_id and edge_indices[1, i] == dst_id:
                val = np.expm1(snapshots[t].edge_attr[i].item()) # Reverse Log
                break
    history_values.append(val)

plt.figure(figsize=(10, 5))
plt.plot(history_times, history_values, marker='o', color='#3498db', linewidth=3, label='Exposure ($M)')

# Highlight the crash point (Label was 1 at T=182, predicting drop at T+1)
plt.axvline(x=182, color='red', linestyle='--', label='Dynamic Model Alert (Rank #1)')
plt.text(182.2, max(history_values)*0.9, 'Contagion Detected', color='red')

plt.xlabel('Time Quarter (t)')
plt.ylabel('Exposure Amount ($ Millions)')
plt.title('Trajectory of "IM -> MD" Risk')
plt.legend()
plt.tight_layout()
plt.savefig('graph_2_trajectory.png')
print("Graph 2 Saved: 'graph_2_trajectory.png'")



k_values = [10, 50, 100, 500, 1000]
precision_dyn = []
precision_static = []

for k in k_values:
    # Dynamic Precision
    top_k_dyn = df_merged.sort_values('Dynamic_Rank').head(k)
    prec_d = top_k_dyn['label'].mean() * 100 # Convert to %
    precision_dyn.append(prec_d)
    
    # Static Precision
    top_k_stat = df_merged.sort_values('Static_Rank').head(k)
    prec_s = top_k_stat['label'].mean() * 100
    precision_static.append(prec_s)

plt.figure(figsize=(10, 6))
plt.plot(k_values, precision_dyn, marker='o', label='Dynamic GNN', color='#2ecc71', linewidth=3)
plt.plot(k_values, precision_static, marker='x', label='Static GNN', color='#e74c3c', linewidth=3)

plt.xlabel('Top K Predictions Checked')
plt.ylabel('Percentage of Actual Crashes Found (%)')
plt.title('Early Warning Precision (Higher is Better)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('graph_3_precision_curve.png')
print("Graph 3 Saved: 'graph_3_precision_curve.png'")
```

```python
import torch
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# 1. Load Data
snapshots = torch.load("sna_graph_snapshots.pt", weights_only=False)
node_map = torch.load("node_mapping.pt")
# Reverse map to get country names
id_to_country = {v: k for k, v in node_map.items()}

# 2. Select the Snapshot: T=182 (The Crash Moment)
TIME_IDX = 182
data = snapshots[TIME_IDX]

# 3. Define the Center of Attention (The Risk Pathway)
SOURCE_NODE = 'IM'
TARGET_NODE = 'MD'
source_id = node_map[SOURCE_NODE]
target_id = node_map[TARGET_NODE]

# 4. Build a NetworkX Graph from the PyTorch Data
G = nx.DiGraph()

# Extract edges
edge_index = data.edge_index.numpy()
edge_attr = data.edge_attr.numpy().flatten()

# Add edges to NetworkX
# (We filter to only show relevant edges to avoid the "Hairball" effect)
# Logic: Include edges that are connected to EITHER Source OR Target
relevant_nodes = set([source_id, target_id])
edges_to_draw = []

for i in range(edge_index.shape[1]):
    u, v = edge_index[0, i], edge_index[1, i]
    weight = np.expm1(edge_attr[i]) # Reverse Log to get Real $
    
    # Filter: Only keep edges connected to our key players
    if u == source_id or v == target_id or u == target_id or v == source_id:
        relevant_nodes.add(u)
        relevant_nodes.add(v)
        edges_to_draw.append((u, v, weight))

# Add nodes and edges to graph
for u in relevant_nodes:
    G.add_node(u, label=id_to_country[u])

for u, v, w in edges_to_draw:
    G.add_edge(u, v, weight=w)

# 5. Visualization Setup
plt.figure(figsize=(12, 8))

# Layout: Spring layout puts connected nodes closer together
pos = nx.spring_layout(G, k=0.5, seed=42) 

# Node Colors
node_colors = []
for node in G.nodes():
    if node == source_id:
        node_colors.append('#e74c3c') # Red (Source/Contagion Origin)
    elif node == target_id:
        node_colors.append('#f39c12') # Orange (Victim)
    else:
        node_colors.append('#95a5a6') # Grey (Neighbors)

# Edge Widths (Proportional to Money)
# We scale them so they are visible but not overwhelming
weights = [G[u][v]['weight'] for u, v in G.edges()]
# Max weight in this subgraph for scaling
max_w = max(weights) if weights else 1
widths = [(w / max_w) * 5 + 0.5 for w in weights]

# Draw!
nx.draw_networkx_nodes(G, pos, node_size=800, node_color=node_colors, alpha=0.9)
nx.draw_networkx_labels(G, pos, labels=nx.get_node_attributes(G, 'label'), font_color='white', font_weight='bold')
nx.draw_networkx_edges(G, pos, width=widths, edge_color='black', alpha=0.6, arrowstyle='->', arrowsize=20)

# Highlight the Specific Risk Edge (IM -> MD) in Red
try:
    nx.draw_networkx_edges(
        G, pos,
        edgelist=[(source_id, target_id)],
        width=4.0,
        edge_color='red',
        connectionstyle="arc3,rad=0.1", # Curve it slightly to make it stand out
        label=f"Risk Flow: {SOURCE_NODE}->{TARGET_NODE}"
    )
except:
    pass # Edge might not exist if value is 0, but usually it exists in T=182

plt.title(f"Contagion Neighborhood: {SOURCE_NODE} -> {TARGET_NODE} at T={TIME_IDX}", fontsize=15)
plt.axis('off') # Hide axis
plt.tight_layout()
plt.savefig('graph_4_network_topology.png')
print("Graph 4 Saved: 'graph_4_network_topology.png'")
plt.show()
```
